# match ... case
- 값의 패턴에 따라 조건을 분기 처리하는 제어문이다.
- python 3.10+ 지원

```python
match 표현식:
    case 값1:
        실행문
    case 값2:
        실행문
    case _:
        기본 실행문
```


````python
def print_drink_price(sel_num: int) -> None:

    match sel_num:
        case 1:
            print("콜라 : 1000원")
        case 2:
            print("사이다 : 900원")

print_drink_price(1)
````

````python
# case _:   확인  -> 기본값(나머지)
def print_drink_price_v2(sel_num: int) -> None:

    match sel_num:
        case 1:
            print('콜라: 1000원')
        case 2:
            print('사이다: 900원')
        case _:
            print('없는 번호 입력됨') # case _: _기본값(나머지)

print_drink_price_v2(1)
print_drink_price_v2(2)
print_drink_price_v2(3)
print_drink_price_v2(44)
````

### 계산기 분기
1. 첫 번째 숫자 입력
2. 산술 연산 기호 입력
3. 두 번째 숫자 입력
4. 결과 출력

단, 산술 연산자가 아닌 기호가 입력되면 에러 발생

````python
def calculate(in_1: int, in_2: int, op: str) -> None:

    match op:
            case "+":
                print(f"{in_1} + {in_2} = {in_1 + in_2}")
            case "-":
                print(f"{in_1} - {in_2} = {in_1 - in_2}")
            case "*":
                print(f"{in_1} * {in_2} = {in_1 * in_2}")
            case "/":
                print(f"{in_1} / {in_2} = {in_1 / in_2}")
            case "%":
                print(f"{in_1} % {in_2} = {in_1 % in_2}")
            case "//":
                print(f"{in_1} // {in_2} = {in_1 // in_2}")
            case "**":
                print(f"{in_1} ** {in_2} = {in_1 ** in_2}")
            case _: # 잘못된 연산자를 입력한 경우
                raise ValueError("잘못된 연산자 입력됨")

input_1 = int(input("첫 번쨰 정수 입력: "))
input_op = input("연산자(+,-,*,/,%,//,**) 입력: ")
input_2 = int(input("두 번쨰 정수 입력: "))

calculate(input_1, input_2, input_op)
````

guard 조건
- case 값 if 조건 형태를 사용하면 단순 값 비교보다 더 세미한 조건을 줄 수 있다.

````python
# 양의 정수, 음의 정수, 0 판독기
def classify_number(n: int) -> None:
    match n:
        case n if n > 0:
            print("양의 정수 입니다.")
        case n if n < 0:
            print("음의 정수 입니다.")
        case _:
            print("0 입니다.")

classify_number(10)
classify_number(-10)
classify_number(0)
````

### tuple 패턴 검사
- `match...case`는 값 하나뿐 아니라 튜플 구조도 검사할 수 있다. 

````python
def inspect_tuple(value: tuple[int, int]) -> None:
    match value:
        case (1, 2):
            print('(1, 2)가 전달 되었습니다.')
        case (1, k) if k < 0:
            print('(1, 음수)가 전달 되었습니다.')
        case (1, _):
            print('(1, 나머지)가 전달 되었습니다.')
        case _:
            print('그 외의 경우가 전달 되었습니다.')

inspect_tuple((1,2))
inspect_tuple((1,-20))
inspect_tuple((1,100))
inspect_tuple((22,33))
````

### dict 패턴 검사
- 딕셔너리도 특정 key와 value 패턴을 기준으로 분기할 수 있다.

````python
def inspect_applicant(applicant: dict[str, object]) -> None:
    match applicant:
        case {'type': 'dev', 'career': career} if career >= 10:
            print('중견 개발자 지원입니다.')
        case {'type': 'dev'}:
            print('개발자 지원입니다.')
        case {'type': 'devops'}:
            print('운영/개발자 지원입니다.')
        case _:
            print('기타 지원입니다.')


applicants:list[dict[str, str | int | None]] = [
    {'type': 'dev', 'career': 15},
    {'type': 'dev', 'career': 5},
    {'type': 'devops', 'career': 0},
    {'type': 'hr'},
]

inspect_applicant(applicants[0])
inspect_applicant(applicants[1])
inspect_applicant(applicants[2])
inspect_applicant(applicants[3])
````
