# dict
- 사전(dictionary) 자료형
- 하나의 요소(item)는 key:value 쌍으로 존재
- 같은 dict 내에서 key는 중복될 수 없다.
    - set는 순서 유지 X, dict는 가능(python 3.7+)
- value는 중복 가능하다.
- key는 hashable 이어야 한다.
    - list, dict, set을 제외한 데이터 타입만 가능하다.
    - 보통 str을 제일 많이 사용
- dict는 index대신 key를 이용해서 value를 조회한다.

````python
# dict 생성
dct = {} # 빈 dict
print(type(dct))
````

````python
dct = {
    "a" : 10,
    "b" : 20,
    "c" : 30
}
print(dct, type(dct))
````

````python
# tuple을 key로 지정
point_map = {(10,20) : "a 좌표"}
print(point_map, type(point_map))
````

````python
# list를 key로 지정
# point_map2 = {[10,20]: "a 좌표"}
# print(point_map2, type(point_map2)) # Error
````

````python
dct = {
    "a" : 10,
    "b" : 20,
    "c" : 30
}
````

````python
# key를 이용한 조회
print("dct['a']:", dct["a"])
print("dct['b']:", dct["b"])
print("dct['c']:", dct["c"])
````

````python
# dict.get(key)를 이용한 조회
print("dct.get('a'):", dct.get("a"))
print("dct.get('b'):", dct.get("b"))
print("dct.get('c'):", dct.get("c"))
````

````python
# [], get() 차이 -> 없는 key를 이용한 조회

# [key] : key가 존재하지 않으면 KeyError 발생
# print("dct['x']:", dct["x"])
````

````python
# get() : key가 존재하지 않으면 None 반환
print("dct['x']:", dct.get("x"))
````

## dict 요소(item == K:V) 추가

````python
students = {
    "s001" : "홍길동",
    "s002" : "신사임당",
    "s003" : "이순신"
}

# students에 item 추가 ([key 역할]= "value 역할")
students["s004"] = "강감찬"

students.update(s005="일론머스크")
# K=V 형식으로 작성 (key는 "" 없이)

students.update({"s006":"빌게이츠", "s007":"백동현"})
# update() 수행 시 dict를 이요해서 여러 item을 한 번에 추가 가능
print(students)
````

## dict 요소(item == K:V) 제거

````python
# pop(key)을 이용한 item 제거
removed = students.pop("s001")
print(removed)
print(students)
````

## dict 내장 함수

````python
# dict() 생성 함수
dct2 = dict(a=10, b=20)
print(dct2)
````

````python
# dict.keys() : dict key만 추출
keys = students.keys()
print(keys, type(keys))
````

````python
# dict.values() : dict에서 values만 추출
values = students.values()
print(values, type(values))
````

````python
# dict.items() : dict의 아이템(key:value)를 tuple로 묶어 반환
items = students.items()
print(items, type(items))
````

## dict unpaking

````python
# dict items unpacking
dct = dict(a=10, b=20, c=30)
print(dct)
````

````python
print(dct.items())

for key, value in dct.items():
    print (key, value)
for a, b in dct.items():
    print (a, b)
````

# dict 복사

### 값의 종류
- 원시 자료형(int, float, bool, ...)
    - 변수에 순수한 값 자체(literal)로 저장되는 자료형 
- 참조 자료형(list, dict, set, str, ...)
    - 변수에 값이 아닌 '값이 저장된 주소'를 저장하는 자료형

### 대입 
- A = B (A 변수에 B 값을 대입) 

```
a = 10 # int(원시 자료형, 값 자체)
b = a # a 에 저장되어 있던 10이라는 값을 복사해서 b에 대입
# a=10, b=10

lst_a = [1,2,3] 
# list(참조 자료형, 값이 저장된 곳의 주소)
lst_b = lst_a 
# list_a에 저장되어있던 주소를 list_b에 복사해서 대입
# lst_a와 lst_b가 같은 주소를 갖는다.
# == 같은 저장 공간을 공유한다.
```
### 얕은 복사(shallow copy)
```
sample = { 'name': '기계식 키보드', 'price': 30\_000, 'origin': 'KR', }

# 얕은 복사 : 주소만 복사하고, 같은 저장 공간을 공유
other1 = sample.copy()
```
### 깊은 복사(deep copy)
```
import copy
other2 = copy.deepcopy(sample)

````python
# 얕은 복사
sample = {
    'name': '기계식 키보드',
    'price': 30_000,
    'origin': 'KR',
}

other1 = sample # sample 변수에 저장된 주소를 other1에 복사'
# sample과 other1은 같은 곳을 바라보게 된다

print("sample:", sample)
print("other1:", other1)

# sample 값 변경
sample['price'] = 35_000

print("sample:", sample)
print("other1:", other1)
````

````python
import copy

nested_sample = {
    'name': '노트북',
    'options': ['16GB RAM', '512GB SSD']
}

other2 = copy.deepcopy(nested_sample)

print("nested_sample:", nested_sample)
print("other2 :", other2)

# nested_sample 변경
nested_sample['name'] = '데스크탑'

print("nested_sample:", nested_sample)
print("other2 :", other2)
````
